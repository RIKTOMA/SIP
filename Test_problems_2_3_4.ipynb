```python
#Problem 2
def f(x):
    r1 = x[0] - 2*x[1] + 5*x[1]**2 - x[1]**3 - 13
    r2 = x[0] - 14*x[1] + x[1]**2 + x[1]**3 - 29
    return r1**2 + r2**2

def grad_f(x):
    r1 = x[0] - 2*x[1] + 5*x[1]**2 - x[1]**3 - 13
    r2 = x[0] - 14*x[1] + x[1]**2 + x[1]**3 - 29
    df_dx0 = 2*(r1 + r2)
    df_dx1 = 2*(r1*(-2 + 10*x[1] - 3*x[1]**2) + r2*(-14 + 2*x[1] + 3*x[1]**2))
    return np.array([df_dx0, df_dx1])

def gcp(x,y):
    return x[0]**2+2*x[1]*y**2+cp.exp(x[0]+x[1])-np.exp(y)

def g(x,y):
    return x[0]**2+2*x[1]*y**2+np.exp(x[0]+x[1])-np.exp(y)

def dg(x,y):
    d1=2*x[0]+np.exp(x[0]+x[1])
    d2=2*y**2+np.exp(x[0]+x[1])
    return np.array([d1,d2])

Y=[(0,1)]
n=2
A=np.zeros((2*n,n))
b=2*np.ones((2*n))
for i in range(n):
    A[i][i]=-1
    A[n+i][i]=1
        
#X=[(linear,[A,b])]
X=[]
y0=np.array([[0.5]])
x0=np.array([1,2])
res1=AlgProxPen(x0,y0,f,gcp,g,X,Y,f=f,df=grad_f,rho=1,tol=10**-3,gamma=0.001,drop_cuts=True)
res2=AlgSGD(x0,y0,f,gcp,f,g,grad_f,dg,X,Y)
```

```python
#Prob 3
def f(x):
    return 1/3*x[0]**2+0.5*x[0]+x[1]**2

def grad_f(x):
    return np.array([2/3*x[0]+0.5, 2*x[1]])

def gcp(x,y):
    return (1-x[0]**2*y**2)**2-x[0]*y**2-x[1]**2+x[1]

def g(x,y):
    return (1-x[0]**2*y**2)**2-x[0]*y**2-x[1]**2+x[1]

def dg(x,y):
    d1=-4*(1-x[0]**2*y**2)*x[0]*y**2-y**2
    d2=-2*x[1]+1
    return np.array([d1,d2])

Y=[(0,1)]
X=[(-10,10),(-10,10)]
y0=np.array([[0.5]])
x0=np.array([-5,-5])
res1=AlgProxPen(x0,y0,f,gcp,g,X,Y,f=f,df=grad_f,rho=1,tol=10**-3,gamma=0.5 ,drop_cuts=True,convex=False)
res2=AlgSGD(x0,y0,f,gcp,f,g,grad_f,dg,X,Y,tol=10**-3)
```

```python
#Problem 4
c=1.5
def f(x):
    return c**2*np.exp(x[0])+np.exp(x[1])

def grad_f(x):
    return np.array([c**2*np.exp(x[0]), np.exp(x[1])])

def gcp(x,y):
    return y-np.exp(x[0]+x[1])

def g(x,y):
    return y-np.exp(x[0]+x[1])

def dg(x,y):
    d1=-np.exp(x[0]+x[1])
    d2=-np.exp(x[0]+x[1])
    return np.array([d1,d2])
Y=[(0,1)]
X=[(-10,10),(-10,10)]
y0=np.array([[0.5]])
x0=np.array([0,0])
res1=AlgProxPen(x0,y0,f,gcp,g,X,Y,f=f,df=grad_f,rho=1,tol=10**-3,gamma=0.1,drop_cuts=True,convex=False)
res2=AlgSGD(x0,y0,f,gcp,f,g,grad_f,dg,X,Y)
```
