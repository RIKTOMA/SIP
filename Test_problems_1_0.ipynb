```python
import numpy as np
import matplotlib.pyplot as plt
import cvxpy as cp
from functions import AlgPenDisc
from functions import AlgProxPen
from functions import linear
```

```python
def test(m,d,seed=0):
    np.random.seed(seed)
    res=np.zeros([5,5,d])
    k=0
    for d in np.arange(0,d):
        Y=[(0,1)]
        n=m
        A=np.zeros((2*n+1,n+1))
        b=np.ones((2*n+1))
        for i in range(n):
            A[i][i]=-1
            A[n+i][i]=1
            A[-1][-1]=-1
            b[-1]=0
        
        X=[(linear,[A,b])]
        X1=[(-1,1) for i in range(n)]
        Y0=[[np.random.rand()]]
        x0=np.zeros(n+1)
        alpha=np.random.rand(n+1)
        beta=np.random.rand(n+1)
        gamma=np.random.rand(n+1)
        delta=np.random.rand(n+1)
        for i in range(n):
            delta[i]=delta[i]/np.sum(delta[:n])
        def fcp(x):
            return x[-1]

        def gcp(x,y):
            return -x[n]+cp.sum([((i+alpha[i])*x[i-1]-(i+beta[i])/n-np.sin(2*np.pi*y[0]+gamma[i]+i))**2  for i in np.arange(1,n+1)])

        def df(x):
            z=np.zeros(np.size(x))
            z[-1]=1
            return(z)

        def dg(x,y):
            z=np.zeros(np.size(x))
            z[-1]=-1
            for i in np.arange(1,n+1):
                z[i-1]=(i+alpha[i])*2*((i+alpha[i])*x[i-1]-(i+beta[i])/n-np.sin(2*np.pi*y+gamma[i]+i))
            #print(z)
            return(z)


        def g(x,y):
            return -x[n]+cp.sum([((i+alpha[i])*x[i-1]-(i+beta[i])/n-np.sin(2*np.pi*y[0]+gamma[i]+i))**2 for i in np.arange(1,n+1)])
        def g1(x,y):
            return -x[n]+cp.sum([((i+alpha[i])*x[i-1]-(i+beta[i])/n-np.sin(2*np.pi*y+gamma[i]+i))**2 for i in np.arange(1,n+1)])

        #Penalty
        rl=AlgPenDisc(x0,Y0,fcp,gcp,g,X,Y,rho=1,drop_cuts=False)
        res[0][4][k]=rl[2]
        res[1][4][k]=rl[3]
        res[2][4][k]=rl[4]
        res[3][4][k]=rl[5]
        res[4][4][k]=rl[-1]

        #Prox
        rl=AlgProxPen(x0,Y0,fcp,gcp,g,X,Y,f=fcp,df=df,rho=1,drop_cuts=False,linearize=False,gamma=1)
        res[0][3][k]=rl[2]
        res[1][3][k]=rl[3]
        res[2][3][k]=rl[4]
        res[3][3][k]=rl[5]
        res[4][3][k]=rl[-1]

        
        #ProxLin
        rl=AlgProxPen(x0,Y0,fcp,gcp,g,X,Y,f=fcp,df=df,rho=1,drop_cuts=False,linearize=True,gamma=1)
        res[0][2][k]=rl[2]
        res[1][2][k]=rl[3]
        res[2][2][k]=rl[4]
        res[3][2][k]=rl[5]
        res[4][2][k]=rl[-1]

        #ProxDrop
        rl=AlgProxPen(x0,Y0,fcp,gcp,g,X,Y,f=fcp,df=df,rho=1,drop_cuts=True,linearize=False,gamma=1)
        res[0][0][k]=rl[2]
        res[1][0][k]=rl[3]
        res[2][0][k]=rl[4]
        res[3][0][k]=rl[5]
        res[4][0][k]=rl[-1]
            
        
        #ProxLinDrop
        rl=AlgProxPen(x0,Y0,fcp,gcp,g,X,Y,f=fcp,df=df,rho=1,drop_cuts=True,linearize=True,gamma=1)
        res[0][1][k]=rl[2]
        res[1][1][k]=rl[3]
        res[2][1][k]=rl[4]
        res[3][1][k]=rl[5]
        res[4][1][k]=rl[-1]

        
        k+=1
    
    
   
    print("exp_prox_sin_dim"+str(m)+"_seed"+str(seed))
    np.save("exp_prox_sin_dim"+str(m)+"_seed"+str(seed), res)
    return res

def tables(res):
    print("Times mean/median ProxDrop",np.mean(res[0][0]),np.median(res[0][0]))
    print("Times mean/median ProxLinDrop",np.mean(res[0][1]),np.median(res[0][1]))
    print("Times mean/median ProxLin",np.mean(res[0][2]),np.median(res[0][2]))
    print("Times mean/median Prox",np.mean(res[0][3]),np.median(res[0][3]))
    print("Times mean/medianPenalty",np.mean(res[0][4]),np.median(res[0][4]))
    
    print("Feasibility mean/median ProxDrop",np.mean(res[1][0]),np.median(res[1][0]))
    print("Feasibility mean/median ProxLinDrop",np.mean(res[1][1]),np.median(res[1][1]))
    print("Feasibility mean/median ProxLin",np.mean(res[1][2]),np.median(res[1][2]))
    print("Feasibility mean/median Prox",np.mean(res[1][3]),np.median(res[1][3]))
    print("Feasibility mean/median Penalty",np.mean(res[1][4]),np.median(res[1][4]))

    print("iterations k mean/median ProxDrop",np.mean(res[3][0]),np.median(res[3][0]))
    print("iterations k mean/median ProxLinDrop",np.mean(res[3][1]),np.median(res[3][1]))
    print("iterations k mean/median ProxLin",np.mean(res[3][2]),np.median(res[3][2]))
    print("iterations k mean/median Prox",np.mean(res[3][3]),np.median(res[3][3]))
    print("iterations k mean/median Penalty",np.mean(res[3][4]),np.median(res[3][4]))
    
    print("Calls vf mean/median ProxDrop",np.mean(res[2][0]),np.median(res[2][0]))
    print("Calls vf mean/median ProxLinDrop",np.mean(res[2][1]),np.median(res[2][1]))
    print("Calls vf mean/median ProxLin",np.mean(res[2][2]),np.median(res[2][2]))
    print("Calls vf mean/median Prox",np.mean(res[2][3]),np.median(res[2][3]))
    print("Calls vf mean/median Penalty",np.mean(res[2][4]),np.median(res[2][4]))

    print("rho incrs  mean/median ProxDrop",np.mean(res[4][0]),np.median(res[4][0]))
    print("rho incrs  mean/median ProxLinDrop",np.mean(res[4][1]),np.median(res[4][1]))
    print("rho incrs  mean/median ProxLin",np.mean(res[4][2]),np.median(res[4][2]))
    print("rho incrs  mean/median Prox",np.mean(res[4][3]),np.median(res[4][3]))
    print("rho incrs  mean/median Penalty",np.mean(res[4][4]),np.median(res[4][4]))

    
    

    plt.legend(["Penalty","Prox", "ProxLin","ProxDrop","ProxLinDrop"])
    labels=["Penalty","Prox", "ProxLin","ProxDrop","ProxLinDrop"]
    d=[res[0][4],res[0][3],res[0][2],res[0][0],res[0][1]]
    plt.boxplot(d,showfliers=False,showmeans=True, meanline=True, tick_labels=labels)
    
```
